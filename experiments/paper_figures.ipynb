```python
import matplotlib.pyplot as plt
from wdn_osp_ll.sensor_placement_optimization import load_or_compute
from wdn_osp_ll.visualization import (
    plot_pareto_front_nsensors_msa,
    plot_parallel_coordinates,
    plot_hypervolume_bar,
    plot_hypervolume_trajectory,
)
```

```python
boatwright_hanoi_cfg = {
    "problem": {
        "leak_analysis": {
            "config": "hyd_analysis/hanoi_base.yaml",
            "emitter_coefficient_filter": None,   # null → all emitter coefficients
            "signature_matrix": "chi_squared",      # sensitivity | chi_squared
            "scale_nodes_by_scenario_mean": True,
        },

        "dd_leak_localization": {
            "name": "two_steps",
            "args": {
                "reconstruction_method": {
                    "name": "boatwright",
                    "args": {
                        "metric": "length",
                        "directed": False,
                        "exponent": None,
                    },
                },
                "search_area_definition_method": {
                    "name": "relative_threshold",
                    "args": {
                        "threshold": None,
                    },
                },
            },
        },

        "excluded_junctions": None,        # null → all junctions
        "n_sensors": None,

        "metrics": [
        ],

        "objectives": [ # subset of metrics, in pygmo order, added automatically to metrics
            "mean_penalized_search_area",
            "n_sensors",
        ],
    },

    "algorithm": {
        "generations": 100,
        "report_every": 10,
        "n_islands": 10,
        "population_size": 100,
    },
}

boatwright_hanoi_res = load_or_compute(config=boatwright_hanoi_cfg)
```
