# Hydraulic Analysis Configuration Tests

We use this notebook for a sanity-check on the Leak Analysis configurations. 
We inspect:
- the leak signature matrices (raw and node-scaled) across emitter coefficients, and
- the distribution of the simulated leak magnitudes relative to the network demand

```python
from wdn_osp_ll.hyd_analysis.data_structures import WDNLeakAnalysis
from wdn_osp_ll.visualization.hyd_analysis import (
    plot_lsms,
    plot_leak_magnitudes,
    plot_relative_leak_magnitudes,
)
```

## Hanoi

```python
hanoi_analysis = WDNLeakAnalysis.load_or_compute("hanoi_base.yaml")
print(hanoi_analysis.n_junctions, hanoi_analysis.n_leak_scenarios, hanoi_analysis.emitter_coefficients)

plot_lsms(hanoi_analysis)
plot_leak_magnitudes(hanoi_analysis)
plot_relative_leak_magnitudes(hanoi_analysis)
```
